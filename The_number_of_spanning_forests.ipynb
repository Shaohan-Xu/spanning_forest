```python
# Define H_1 graph and compute Tutte polynomial at (2,1) (number of spanning forests)
from sage.plot.text import text
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Circle

# 15. H_1
G15 = Graph(multiedges=True, name="H_1")
G15.add_edges([
    (1,4), (1,2), (2,3), (3,4), (1,3), (2,4),  # Edges of the outer square
    ("v\u2081","v\u2084"), ("v\u2081","v\u2082"), ("v\u2082","v\u2083"), ("v\u2083","v\u2084"),  # Edges of the inner square
    ("v\u2081","v\u2083"), ("v\u2082","v\u2084"),  # Diagonals of the inner square
    (1,"v\u2081"), (2,"v\u2082"), (3,"v\u2083"), (4,"v\u2084")  # Connecting edges between squares
])

# Vertex positions based on the image
H1_pos = {
    "v\u2081":(-1,2),
    "v\u2084":(1,2),
     "v\u2083":(1,0),
     "v\u2082":(-1,0),
    1:(-2,3),
    4:(2,3),
    3:(2,-1),
    2:(-2,-1),
}

# Compute Tutte polynomial value at (2,1)
T = G15.tutte_polynomial()
T21 = T(2, 1)

# Display computation results
graph_name = G15.name()
print(f"Graph name: {graph_name}")
print(f"The number of spanning forests: T_{{{graph_name}}}(2,1) = {T21}")

# Manual drawing with curved edges
def draw_curved_edge(ax, start, end, curvature=0.5, color='black', linewidth=2):
    """Draw a curved edge"""
    # Calculate midpoint
    mid_x = (start[0] + end[0]) / 2
    mid_y = (start[1] + end[1]) / 2
    
    # Calculate perpendicular direction
    dx = end[0] - start[0]
    dy = end[1] - start[1]
    length = np.sqrt(dx*dx + dy*dy)
    
    if length > 0:
        # Perpendicular direction
        perp_x = -dy / length
        perp_y = dx / length
        
        # Control point position
        control_x = mid_x + perp_x * curvature * length
        control_y = mid_y + perp_y * curvature * length
        
        # Generate quadratic Bezier curve
        t = np.linspace(0, 1, 100)
        curve_x = (1-t)**2 * start[0] + 2*(1-t)*t * control_x + t**2 * end[0]
        curve_y = (1-t)**2 * start[1] + 2*(1-t)*t * control_y + t**2 * end[1]
        
        ax.plot(curve_x, curve_y, color=color, linewidth=linewidth, zorder=1)

# Create manually drawn figure
fig, ax = plt.subplots(figsize=(6, 6))

# Draw vertices
vertices = list(H1_pos.keys())
for vertex, pos in H1_pos.items():
    circle = Circle((pos[0], pos[1]), 0.15, fill=True, color='blue', zorder=5)
    ax.add_patch(circle)
    ax.text(pos[0], pos[1], vertex, ha='center', va='center', 
            color='white', fontsize=12, weight='bold', zorder=6)

# Draw straight edges (except for the curved ones)
straight_edges = [
    (1,4), (1,2), (2,3), (3,4),  # Outer square edges
    ("v\u2081","v\u2084"), ("v\u2081","v\u2082"), ("v\u2082","v\u2083"), ("v\u2083","v\u2084"),  # Inner square edges
    ("v\u2081","v\u2083"), ("v\u2082","v\u2084"),  # Inner square diagonals
    (1,"v\u2081"), (2,"v\u2082"), (3,"v\u2083"), (4,"v\u2084")  # Connecting edges
]

for edge in straight_edges:
    start = H1_pos[edge[0]]
    end = H1_pos[edge[1]]
    ax.plot([start[0], end[0]], [start[1], end[1]], 'k-', linewidth=2, zorder=2)

# Draw curved edges (1,3) and (2,4)
curved_edges = [(1,3), (2,4)]
curvatures = [0.3, 0.3]  # Opposite curvature directions for the two edges

for i, edge in enumerate(curved_edges):
    start = H1_pos[edge[0]]
    end = H1_pos[edge[1]]
    draw_curved_edge(ax, start, end, curvature=curvatures[i], linewidth=1.5)  # Reduced line width

# Set figure properties with tighter bounds
# Calculate the bounds based on vertex positions
x_coords = [pos[0] for pos in H1_pos.values()]
y_coords = [pos[1] for pos in H1_pos.values()]

# Add a small margin (10%) around the vertices
x_margin = (max(x_coords) - min(x_coords)) * 0.1
y_margin = (max(y_coords) - min(y_coords)) * 0.1

ax.set_xlim(min(x_coords) - x_margin, max(x_coords) + x_margin)
ax.set_ylim(min(y_coords) - y_margin, max(y_coords) + y_margin)
ax.set_aspect('equal')
ax.axis('off')

# Remove extra padding around the plot
plt.tight_layout(pad=0.5)  # Reduced padding

plt.show()
```
